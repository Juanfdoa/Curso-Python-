Es el proceso de inspeccionar, limpiar, transformar y modelar datos para extraer informacion significativa y apoyar la toma de decisiones.

### Herramientas utiles para el analisis de datos en python
- Pandas
- MatplotLib

### Carga y Exploración

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Sales data
data = pd.read_csv("sales_data.csv")
```

```python
# first 5 records
print(data.head())
```

```python
# Summary of Dataset
print(data.info())
```

```python
# Statistical Summary
print(data.describe())
```

```python
# Check for Missing Values
print(data.isnull().sum())
```

### Limpieza y Transformación

```python
# Fill Missing Values
data["Product_Category"] = data["Product_Category"].fillna("Unknown")
```

```python
# Drop Rows with missing values
data = data.dropna()
```

```python
# Convert Date Column to Datetime
data["Date"] = pd.to_datetime(data['Date'])
```

```python
# Convert values to numerical values
data["Sales_Amount"] = pd.to_numeric(data['Sales_Amount'], errors='coerce')
```

```python
# Add a Year-Montn Column
data["Year_Month"] = data['Date'].dt.to_period('M')
```

```python
# Add a Revenue Column (if Quantity and Price are available)
data["Revenue"] = data['Quantity'] * data['Price']
```

### Generar visualizaciones

```python
# Total sales by Month
montly_sales = data.groupby("Year_Month")["Sales_Amount"].sum()
print(montly_sales)
```

```python
# Top Products
top_products = data.groupby("Product_Name")["Revenue"].sum().sort_values(ascending=False).head(5)
print(top_products)
```

```python
# Plot Montly Sales
montly_sales.plot(kind="bar", figsize=(10, 6), color="skyblue")
plt.title("Montly Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

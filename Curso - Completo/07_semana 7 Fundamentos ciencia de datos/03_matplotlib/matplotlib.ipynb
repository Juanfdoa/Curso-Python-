Matplotlib es una de las librerías más usadas en Python para crear gráficos y visualizaciones de datos.

### ¿Para qué sirve?
Con Matplotlib puedes crear:
- Gráficos de líneas 📈
- Gráficos de barras 📊
- Histogramas
- Gráficos de dispersión (scatter)
- Subplots (varios gráficos en uno)

### Gráfico de líneas

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Plot
plt.plot(x, y, label="Line")
plt.title("Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

### Gráfico de barras

```python
import matplotlib.pyplot as plt

# Data
categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 30]

# Plot
plt.bar(categories, values, color="skyblue")
plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

### Gráfico de dispersión

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2.5, 3.7, 4.6, 8.0, 10.5]

# Plot
plt.scatter(x, y, color="red")
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

### Gráfico personalizado

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Plot
plt.plot(x, y, label="Line", color="green", linestyle="--", marker="o")
plt.title("Customized Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.annotate("Max Value", xy=(5, 10), xytext=(4, 8), arrowprops=dict(facecolor="black", arrowstyle="->"))         
plt.legend()
plt.show()
```

### Subgráfico

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 2, 3, 4, 5]

# Subplots
plt.subplot(1, 2, 1)
plt.plot(x, y1, color="blue")
plt.title("Graph 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2, color="red")
plt.title("Graph 2")

plt.tight_layout()
plt.show()
```

### Gráficos desde archivos CSV

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load Data
data = pd.read_csv("data.csv")

# Plot
plt.plot(data["X"], data["Y"], label="Data from CSV")
plt.title("Plot from CSV")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

### Ejercicio
- Graph Plotter

```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_graph():
    print("\nWelcome to the Graph Plotter!")
    print("Choose a graph type: ")
    print("1. Line Graph")
    print("2. Bar Chart")
    print("3. Scatter Plot")

    choice = input("Enter the number of your choice: ")
    if choice not in ["1", "2", "3"]:
        print("Invalid choice. Exiting.")
        return
    
    print("Chose data input method:")
    print("1. Enter data manually")
    print("2. Load data from CSV file")

    data_choice = input("Enter the number of your choice: ")

    if data_choice == "1":
        x = list(map(float, input("Enter X values separated by spaces: ").split()))
        y = list(map(float, input("Enter Y values separated by spaces: ").split()))
    elif data_choice == "2":
        file_path = input("Enter the path to the CSV File")
        try:
            data = pd.read_csv(file_path)
            x = data.iloc[:, 0]
            y = data.iloc[:, 1]
        except Exception as e:
            print("Error loading CSV file:", e)
            return
    else:
        print("Invalid choice. Exiting.")

    if choice == "1":
        plt.plot(x, y, label="Line Graph", marker="o")
    elif choice == "2":
        plt.bar(x, y, color="skyblue")
    elif choice == "3":
        plt.scatter(x, y, color="red")

    plt.title("Graph Plotter")
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.legend()
    plt.grid(True)

    save_choice = input("Do you want to save the graph? (yes/no): ").lower()
    if save_choice == "yes":
        file_name = input("Enter the file name:")
        plt.savefig(f"{file_name}.png")
        print(f"Graph saved as {file_name}")
    else:
        plt.show()

if __name__ == "__main__":
    plot_graph()
```

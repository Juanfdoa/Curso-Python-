- Constructores: Es un metodo especial utilizado para inicializar objectos 

```python
# Constructor
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    # Metodo de instancia
    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")

# Object
book1 = Book('1984','George Orwell')
book1.display_info()
```

### Metodos de Clase y Metodos estaticos
- Metodos de clase: Son métodos que pertenecen a la clase y no a una instancia específica.
- Metodos staticos: Son métodos que también pertenecen a la clase, pero no reciben ni self ni cls.

```python
class Utility:
    app_version = "1.0"

    @classmethod
    def get_version(cls):
        print(f"App Version: {cls.app_version}")

    @staticmethod
    def greet():
        print("Hello! Welcome to the app.")

Utility.get_version()
Utility.greet()
```

### Encapsulacion
Es el principio de ocultar los datos internos de una clase y permitir su acceso solo a través de métodos controlados.


```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance     #Propiedad privada o encapsulada

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}, New balance: ${self.__balance}")
        else:
            "Invalid deposit amount"

    def get_balance(self):
        return self.__balance
    
account = Account('Jhon Doe',1000)
account.deposit(500)
print(f"Account balance: ${account.get_balance()}")
```

### Ejercicio
- Library Management System

```python
import json

class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author
        self.is_borrowed = False

    def display_info(self):
        status = "Available" if not self.is_borrowed else "Borrowed"
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Status: {status}\n")

class Library:
    def __init__(self):
        self.books = []

    def add_book(self,title,author):
        new_book = Book(title,author)
        self.books.append(new_book)
        print(f"Book '{title}' by {author} added to the library.")

    def view_books(self):
        if not self.books:
            print("No books in the library")
        else:
            print("\n----- Library Books -----")
            for book in self.books:
                book.display_info()

    def borrow_book(self, title):
        for book in self.books:
            if book.title == title and not book.is_borrowed:
                book.is_borrowed = True
                print(f"Book '{title}' has been borrowed. Enjoy reading")
                return
        print(f"Book '{title}' is not available for borrowing.")

    def return_book(self,title):
        for book in self.books:
            if book.title == title and book.is_borrowed:
                book.is_borrowed = False
                print(f"Book '{title}' has been returned.")
                return
        print(f"Book '{title}' is not in the library.")

    def save_to_json(self, filename="library_02.json"):
        data = []
        for book in self.books:
            data.append({
                "title": book.title,
                "author": book.author,
                "is_borrowed": book.is_borrowed
            })

        with open(filename, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=4)
        print("Library saved successfully.")


    def load_from_json(self, filename="library_02.json"):
        try:
            with open(filename, "r", encoding="utf-8") as f:
                data = json.load(f)

            self.books = []  # limpia la lista actual

            for item in data:
                book = Book(item["title"], item["author"])
                book.is_borrowed = item["is_borrowed"]
                self.books.append(book)
            print("Library loaded successfully.")

        except FileNotFoundError:
            print("No saved library file found.")

# Main Program
library = Library()

while True:
    print("\n------ Library Management System ------")
    print("1. Add Book")
    print("2. View Books")
    print("3. Borrow Book")
    print("4. Return Book")
    print("5. Save Books on Json File")
    print("6. Load Books from json File")
    print("7. Exit")

    choice = input('Enter your choice (1-7): ').strip()
    if choice == "1":
        title = input('Enter book title: ').strip()
        author = input('Enter author name: ').strip()
        library.add_book(title,author)
    elif choice == "2":
        library.view_books()
    elif choice == "3":
        title = input('Enter book title to borrow: ').strip()
        library.borrow_book(title)
    elif choice == "4":
        title = input('Enter book title to return: ').strip()
        library.return_book(title)
    elif choice == "5":
        library.save_to_json()
    elif choice == "6":
        library.load_from_json()
    elif choice == "7":
        print("Exiting the Library Management System. Goodbye!")
        break
    else:
        print("Invalid choice. Please select a valid option (1-7)")
```

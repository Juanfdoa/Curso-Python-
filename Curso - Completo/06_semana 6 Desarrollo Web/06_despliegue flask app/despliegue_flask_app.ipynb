### Preparación
- Tener bien planteada la estructura de carpetas (templates,estatic, etc)
- pip freeze > requirements.txt (Para crear el archivo requierements.txt y agregarle las bibiotecas basicas necesarias para la app flask)
- agregar gunicorn en el archivo requirements.txt
- crear archivo "Procfile" en la raiz de proyecto y agregarle (web: gunicorn app:app)

### Despliegue en Heroku
- Ir a "Heroku cli" y descargar heroku
- heroku --version (confirmar instalacion)
- heroku login (para iniciar session o crear cuenta)
- En la terminal en la que esta logueado, dirigirse a la app flask
- heroku create my-heroku-first-test-app (para crear la aplicacion y retorna la url de la app)
- Inicializar git y enlazarlo con el repositorio github que heroku creo (Lo encuentras en el panel de control de heroku)
- git init 
- git add . 
- git commit -m "Initial Commit"
- git remote add heroku https://git.heroku.com/my-heroku-first-test-app.git
- git push heroku main
- heroku open

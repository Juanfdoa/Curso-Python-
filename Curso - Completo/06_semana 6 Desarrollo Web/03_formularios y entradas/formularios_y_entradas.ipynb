### Flask-WTF
Flask-WTF es una extensión de Flask que facilita la creación y validación de formularios en aplicaciones web.
Es básicamente una integración entre:
- Flask
- WTForms (librería para manejar formularios en Python)

### Instalacion
- pip install flask-wtf
- pip show flask-wtf
- pip install email-validator (Validador)

### Estructura

```python
from flask import Flask, render_template, request, flash, redirect, url_for
from flask_wtf import FlaskForm
from wtforms import StringField, TextAreaField, SubmitField
from wtforms.validators import DataRequired, Email

app = Flask(__name__)
app.secret_key = 'Your_secret_key'

# Contact Form Class
class ContactForm(FlaskForm):
    name = StringField("Your Name", validators=[DataRequired()])
    email = StringField("Email", validators=[DataRequired(), Email()])
    message = TextAreaField("Message", validators=[DataRequired()])
    submit = SubmitField("Submit")

@app.route("/", methods=["GET", "POST"])
def contact():
    form = ContactForm()
    if form.validate_on_submit():
        flash(f"Message sent successfully from {form.name.data}!", "success")
        return redirect(url_for('contact'))
    return render_template("contact.html", form=form)

if __name__ == "__main__":
    app.run(debug=True)
```

 APIS REST
</h1>

Una API REST (Representational State Transfer) es una forma estándar de comunicar aplicaciones a través de internet usando HTTP. Básicamente permite que un sistema le pida datos o envíe datos a otro sistema.

En otras palabras:
Una API REST es una interfaz para que programas hablen entre sí.

### Metodós
- Get -> Recupera datos
- Post -> Agrega nuevos datos
- Put -> Actualizacion de datos existentes
- Delete -> Elimina datos

### Configuracion Api Rest con Flask
- pip install flask flask-restful

### Estructura

```python
from flask import Flask, jsonify

app = Flask(__name__)

# Root Endpoint
@app.route('/')
def home():
    return jsonify({"message":"welcome to the Mini Weather API!"})

if __name__ == '__main__':
    app.run(debug=True)
```

<h1 style="text-align:center">
 Flask
</h1>

Flask es un microframework web escrito en Python que se usa para crear aplicaciones web de forma rápida y sencilla.

Es “micro” porque incluye solo lo esencial para que una aplicación web funcione, pero permite agregar fácilmente extensiones cuando necesitas más funcionalidades (base de datos, autenticación, formularios, etc.).

### ¿Para qué sirve Flask?
- Crear APIs REST
- Construir aplicaciones web
- Hacer backends para apps móviles
- Crear microservicios
- Desarrollar proyectos pequeños y medianos rápidamente

### Configuración
- pip install flask
- python -m flask --version

### Estructura Basica

```python
from flask import Flask

# Create Flask App
app = Flask(__name__)

# Define Route
@app.route("/")
def home():
    return "Hola Mundo"

# Run App
if __name__ == "__main__":
    app.run(debug=True)
```
